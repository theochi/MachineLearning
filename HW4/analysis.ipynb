1. How do the feature types differ, and why does scaling matter more for SVMs than Decision Trees?

The features differ because this dataset mixes two kinds of information instead of using only one kind of feature. The continuous features are real measurements whereas, the categorical features are labels. Scaling is critical for the SVM because the RBF kernel works by measuring the distance between patients. Without scaling, cholesterol would completely dominate that distance, and smaller-range features like oldpeak (0–6) would barely count even though they are strong predictors. Decision Trees don’t need scaling because they split on one feature at a time using a threshold, like chol <= 240. Rescaling a feature changes where the threshold sits but not which patients end up on each side, so the tree’s results stay the same.


2. How do hidden missing values affect medical data, and why use different imputation strategies?


Hidden placeholders like ? aren’t recognized as missing by default, so the data looks complete when it isn’t. In this dataset, ca and thal were read as text instead of numbers, which would break or distort any calculation on those columns until the ? values were converted to NaN. In medical data, a missing value is often missing for a reason. Simply ignoring it can bias the model, and deleting those patients throws away information from a dataset that only has 303 people - this is an issue. Using a median for vitals like cholesterol or blood pressure is better. Using a mode for ca and thal is also better. A mean could produce a category that doesn’t exist whilst the mode always fills in a real category.


3. How do an RBF SVC and a Decision Tree build their decision boundaries, and what are the interpretability tradeoffs?

An RBF SVC and a Decision Tree build their decision boundaries in very different ways. The RBF SVC maps patients into a higher-dimensional space and finds the boundary with the widest possible margin between the disease and no-disease groups.A Decision Tree splits the data on one feature at a time, picking the split that best separates the two groups each time. This divides the feature space into rectangular, box-shaped regions that get one prediction each. This creates a tradeoff in interpretability. The SVC performed better in this project (ROC-AUC of 0.963 vs 0.873 for the tree). The Decision Tree is much easier to interpret because a doctor can follow its if/then rules from the top of the tree to a prediction.

4. How do C and gamma affect bias, variance and generalization?

C controls how strongly the model is penalized for misclassifying training patients. A low C allows more mistakes and higher bias. A high C tries to classify every training patient correctly, which creates a tighter boundary. This means higher variance and risk of overfitting). gamma controls how far each patient’s influence reaches. A low gamma creates a smooth, broad boundary. A high gamma makes each patient’s influence very local, so the boundary bends around individual points and starts memorizing noise.For medical data, generalization matters more than training accuracy because the model has to work on new patients.
