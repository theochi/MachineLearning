2.

Hard voting gives each model one vote for a class label and predicts the majority. Soft voting averages the predicted probabilities and picks the class with the highest average. Both resulted in a 72.7% test accuracy but had different errors. Hard voting caught more diabetic cases while soft voting was able to predict diabetic cases better.
Calibration matters because averaging only works if a probability of 0.8 means the same thing from each model. Logistic Regression produces probabilities directly. SVC has no native probabilities. The decision tree reports leaf class proportions, which tend to be overconfident. 23% of its test predictions were below 0.05 or above 0.95, so it can outweigh the other two in the average. Based on the data one can infer that voting helps most when the base models are similarly strong and make different mistakes.

OOB is efficient because it comes free from a single fit. OOB uses all 614 training rows without setting any aside, which matters for a small dataset. Unfortunately though, OOB prediction uses only about a third of the trees. This means its score can come out a little lower than the model’s true accuracy.

4.

In Bagging, every tree can choose from all 8 features at each decision. Since glucose is by far the most useful feature, most trees end up looking very similar. Random Forest only lets each tree choose from 2 random features at each decision, so trees are forced to use other features like BMI and age. Different trees make different mistakes, and when you average them, those mistakes cancel out more, so predictions become more stable. Random Forest and Bagging scored almost the same in the results.

5.

AdaBoost builds simple trees one at a time. After each stump, the patients it got wrong are given more importance, so the next stump focuses on fixing those mistakes. AdaBoost showed very little overfitting. It got 80.9% on training data and 76.0% on test data. Bagging and Random Forest, by comparison, scored 100% on training but only about 74% on test. To keep overfitting low, one could lower the learning rate so each stump changes the model less. You could also keep the trees as simple stumps, and pick the number of stumps using a validation set rather than the test set.
