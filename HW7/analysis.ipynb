1. Feature scaling is essential because the raw features are on very different scales. Proline runs from 278 to 1,680; std. deviation 314.9, magnesium from 70 to 162; std. deviation 14.3, while hue: 0.48–1.71; std. deviation 0.229 and nonflavanoid phenols: 0.13–0.66; std. deviation 0.124 varies by less than 1.5 units. Because K-Means, DBSCAN and HDBSCAN all use Euclidean distance, proline alone makes up 99.8% of the total raw variance - magnesium 0.2%, everything else around 0%. Unscaled distance is mainly proline. In this data it makes a big difference. Unscaled K-Means (k=3) gets an Adjusted Rand Index of only 0.371 against the true cultivars. After StandardScaler it reaches 0.897. The boxplots in Task 1 show all 13 features sitting on one z-score scale after scaling.

2. Both metrics chose k = 3, which matches the 3 true cultivars. Inertia fell sharply from 2,314 to 1,659 to 1,278 (k=1 - 3), then flattened (k=4 only dropped it about 100 more, to 1,175). The silhouette peaked at 0.285 at k=3. There were discrepancies observed; the silhouette peak was weak: k=2 (0.259) and k=4 (0.260) were nearly as good. There was a second bump at k=6 (0.237) and the elbow is also a gradual bend rather than a sharp corner, so k=3 only came out on top by a small margin. The low absolute silhouette (0.285, where 1.0 is perfect) means the cultivars overlap in chemical space. This shows up in the K-Means results: 6 of the 71 class 1 wines were placed with the other two cultivars (3 into each).

3. K-Means assigns every wine to its nearest centroid, so it had 0 noise points. It produced 3 clusters (65/51/62) that matched the cultivars closely. DBSCAN (eps=2.4, min_samples=5) found only 2 clusters plus 36 noise points. HDBSCAN found 2 clusters plus 58 noise points. Both merged cultivars 0 and 1 into one large cluster (DBSCAN 57 + 42, HDBSCAN 53 + 33). Wines were labeled noise when they didn’t have enough close neighbors in 13-D space. These were mostly class 1 wines (28 of 36 DBSCAN noise points, 37 of 58 for HDBSCAN). Class 1 is the most spread-out cultivar and lies between the other two (see the middle region of the PCA plots), so its points are too sparse to count as dense.

4. DBSCAN uses a single global eps, and the results were very sensitive to it. eps=0.5 labeled all 178 wines as noise. eps=2.0 / min_samples=5 gave 5 clusters with 85 noise. eps=2.2 / min_samples=3 gave 3 clusters with 44 noise (one cluster had only 3 points). eps=2.6 / min_samples=5 collapsed to 1 cluster. There was no single eps that suited both the dense and the sparse regions. HDBSCAN removes eps altogether. It builds a single-linkage tree over mutual-reachability distances and condenses it by dropping branches with fewer than 5 points; keeping the branches that stay intact longest. In effect, each cluster gets its own density cut. With only min_cluster_size=5,  it found the same 2-cluster structure that DBSCAN needed a grid search to find.

5. Based on my comparative classification experiment the K-Means labels were by far the most useful: 97.2% accuracy (35 of 36 test wines correct). DBSCAN reached 77.8% (8 wrong) and HDBSCAN 75.0% (9 wrong). With only 2 clusters, the density methods put most cultivars 0 and 1 in one cluster. The classifier can only map that cluster to one class, so many wines from the other cultivar are guaranteed to be wrong. The noise category is a mixed bag as well so it mostly just predicts class 1.The internal metric and the real-world result disagreed. DBSCAN and HDBSCAN had higher silhouette scores (0.327 and 0.352 vs 0.285). This was because their noise points were left out of the calculation. The 36-wine test set is also small: one misclassified wine changes accuracy by about 2.8 points.
