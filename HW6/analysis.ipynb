2. The first eigenfaces capture the biggest differences across all the photos. Eigenface 1 & 2 mostly encode overall brightness and lighting direction, and the next ones pick out major features such as the mouth and smile. Later components look more like fine-grained pixel noise. Once the large, shared facial structure is used up, what remains are tiny, image-specific details and pixel noise.

3. It is critical to fit the PCA transformer on the training split so the full dataset lets the test images shape the features the classifier is evaluated on. The results were as follows: PCA on all 400 images chose 123 components for 95% variance. However it fitted on the 320 training images it chose 111, so the test set would have changed the feature space itself. Leakage like this makes test accuracy look better than it would be on new faces.
