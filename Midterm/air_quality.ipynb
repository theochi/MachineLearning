***Task 4***

Identify and handle implicit missing data values

```python
import numpy as np

from sklearn.impute import SimpleImputer

clean_data = df.copy()
numeric_columns = clean_data.select_dtypes(include="number").columns.tolist()
sentinel_counts = (clean_data[numeric_columns] == -200).sum()
display(sentinel_counts.to_frame("Count of -200 values"))
clean_data[numeric_columns] = clean_data[numeric_columns].replace(-200, np.nan)
clean_data = clean_data.dropna(how="all")
display(clean_data.isna().sum().to_frame("Missing values"))

time_text = clean_data["Time"].astype(str).str.strip().str.replace(
    ".", ":", regex=False
)

clean_data["Timestamp"] = pd.to_datetime(
    clean_data["Date"].astype(str).str.strip() + " " + time_text,
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

if clean_data["Timestamp"].isna().any():
    display(
        clean_data.loc[
            clean_data["Timestamp"].isna(), ["Date", "Time"]
        ].head(10)
    )
    raise ValueError("The displayed dates/times could not be converted.")
clean_data = clean_data.sort_values("Timestamp").reset_index(drop=True)

target_column = "CO(GT)"
model_data = clean_data.dropna(subset=[target_column]).copy()
target_values = model_data[target_column].copy()

reference_columns = ["CO(GT)", "NMHC(GT)", "C6H6(GT)", "NOx(GT)", "NO2(GT)"]
feature_data = model_data.drop(columns=reference_columns + ["Date", "Time", "Timestamp"])
feature_data["Hour"] = model_data["Timestamp"].dt.hour
feature_data["Month"] = model_data["Timestamp"].dt.month.astype(str)
feature_data["Weekday"] = model_data["Timestamp"].dt.day_name()

split_point = int(len(feature_data) * 0.8)
train_features = feature_data.iloc[:split_point].copy()
test_features = feature_data.iloc[split_point:].copy()
train_target = target_values.iloc[:split_point].copy()
test_target = target_values.iloc[split_point:].copy()
validation_point = int(len(train_features) * 0.8)
fit_features = train_features.iloc[:validation_point].copy()
validation_features = train_features.iloc[validation_point:].copy()
fit_target = train_target.iloc[:validation_point].copy()
validation_target = train_target.iloc[validation_point:].copy()

empty_columns = fit_features.columns[fit_features.isna().all()].tolist()
fit_features = fit_features.drop(columns=empty_columns)
validation_features = validation_features.drop(columns=empty_columns)
train_features = train_features.drop(columns=empty_columns)
test_features = test_features.drop(columns=empty_columns)

numeric_columns = fit_features.select_dtypes(include="number").columns.tolist()
demo_imputer = SimpleImputer(strategy="median")
imputed_preview = pd.DataFrame(
    demo_imputer.fit_transform(fit_features[numeric_columns]),
    columns=numeric_columns, index=fit_features.index
)
display(imputed_preview.head())
```

***Task 5***

Isolate Numerical and Categorical Features

```python
from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

numeric_columns = train_features.select_dtypes(
    include="number"
).columns.tolist()

categorical_columns = train_features.select_dtypes(
    exclude="number"
).columns.tolist()

def make_preprocessor(scaler=None):
    numeric_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        (
            "scaler",
            StandardScaler() if scaler is None else clone(scaler)
        )
    ])
    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])
    return ColumnTransformer([
        ("numeric", numeric_pipeline, numeric_columns),
        ("categorical", categorical_pipeline, categorical_columns)
    ])

preprocessor = make_preprocessor()
display(preprocessor)
```

***Task 6***

Train Classification Models

```python
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plot

from sklearn.exceptions import ConvergenceWarning
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay,
    log_loss
)

bin_edges = [-np.inf, 1.5, 3.0, np.inf]

fit_classes = pd.cut(
    fit_target, bins=bin_edges, labels=[0, 1, 2], right=False
).astype(int)

validation_classes = pd.cut(
    validation_target, bins=bin_edges, labels=[0, 1, 2], right=False
).astype(int)

train_classes = pd.cut(
    train_target, bins=bin_edges, labels=[0, 1, 2], right=False
).astype(int)

test_classes = pd.cut(
    test_target, bins=bin_edges, labels=[0, 1, 2], right=False
).astype(int)

display(
    train_classes.value_counts()
    .sort_index()
    .to_frame("Training class counts")
)

if set(fit_classes.unique()) != {0, 1, 2}:
    raise ValueError(
        "The fitting period must contain all three classes."
    )

classification_models = {
    "Softmax logistic": LogisticRegression(
        solver="lbfgs",
        max_iter=3000,
        C=1.0
    ),
    "SGD logistic L2": SGDClassifier(
        loss="log_loss",
        penalty="l2",
        alpha=0.0001,
        max_iter=3000,
        tol=0.001,
        random_state=42
    ),
    "SGD logistic stronger L2": SGDClassifier(
        loss="log_loss",
        penalty="l2",
        alpha=0.01,
        max_iter=3000,
        tol=0.001,
        random_state=42
    ),
    "SGD hinge L2": SGDClassifier(
        loss="hinge",
        penalty="l2",
        alpha=0.0001,
        max_iter=3000,
        tol=0.001,
        random_state=42
    )
}

classification_rows = []

for model_name, estimator in classification_models.items():
    model_pipeline = Pipeline([
        ("prepare", make_preprocessor()),
        ("model", clone(estimator))
    ])

    start_time = time.perf_counter()

    with warnings.catch_warnings(record=True) as warning_list:
        warnings.simplefilter("always", ConvergenceWarning)
        model_pipeline.fit(fit_features, fit_classes)

    elapsed_time = time.perf_counter() - start_time
    predictions = model_pipeline.predict(validation_features)

    classification_rows.append({
        "Model": model_name,
        "Validation accuracy": accuracy_score(
            validation_classes, predictions
        ),
        "Validation macro F1": f1_score(
            validation_classes,
            predictions,
            labels=[0, 1, 2],
            average="macro",
            zero_division=0
        ),
        "Fit seconds": elapsed_time,
        "Iterations": int(
            np.max(model_pipeline.named_steps["model"].n_iter_)
        ),
        "Convergence warning": any(
            issubclass(item.category, ConvergenceWarning)
            for item in warning_list
        )
    })

classification_results = pd.DataFrame(
    classification_rows
).sort_values("Validation macro F1", ascending=False)

display(classification_results)

best_classification_name = classification_results.iloc[0]["Model"]

best_classifier = Pipeline([
    ("prepare", make_preprocessor()),
    (
        "model",
        clone(classification_models[best_classification_name])
    )
])

best_classifier.fit(train_features, train_classes)
test_predictions = best_classifier.predict(test_features)

print(
    "Selected using validation macro F1:",
    best_classification_name
)
print(
    "Test accuracy:",
    accuracy_score(test_classes, test_predictions)
)
print(
    "Test macro F1:",
    f1_score(
        test_classes,
        test_predictions,
        labels=[0, 1, 2],
        average="macro",
        zero_division=0
    )
)
print(
    classification_report(
        test_classes,
        test_predictions,
        labels=[0, 1, 2],
        target_names=["Low", "Moderate", "High"],
        zero_division=0
    )
)
ConfusionMatrixDisplay.from_predictions(
    test_classes,
    test_predictions,
    labels=[0, 1, 2],
    display_labels=["Low", "Moderate", "High"]
)
plot.show()

scaler_options = {
    "StandardScaler": StandardScaler(),
    "MinMaxScaler": MinMaxScaler()
}

scaler_results = []
scaler_loss_curves = {}

for scaler_name, scaler in scaler_options.items():
    comparison_preprocessor = make_preprocessor(scaler)

    comparison_fit = comparison_preprocessor.fit_transform(
        fit_features
    )
    comparison_validation = comparison_preprocessor.transform(
        validation_features
    )

    comparison_classifier = SGDClassifier(
        loss="log_loss",
        penalty="l2",
        alpha=0.0001,
        learning_rate="constant",
        eta0=0.01,
        random_state=42
    )

    comparison_losses = []
    start_time = time.perf_counter()

    for epoch in range(30):
        comparison_classifier.partial_fit(
            comparison_fit,
            fit_classes,
            classes=np.array([0, 1, 2])
        )

        comparison_probabilities = (
            comparison_classifier.predict_proba(
                comparison_validation
            )
        )

        comparison_losses.append(
            log_loss(
                validation_classes,
                comparison_probabilities,
                labels=[0, 1, 2]
            )
        )

    elapsed_time = time.perf_counter() - start_time

    comparison_predictions = comparison_classifier.predict(
        comparison_validation
    )

    scaler_results.append({
        "Scaler": scaler_name,
        "Training and evaluation seconds": elapsed_time,
        "Epochs": 30,
        "First validation log loss": comparison_losses[0],
        "Final validation log loss": comparison_losses[-1],
        "Validation accuracy": accuracy_score(
            validation_classes, comparison_predictions
        ),
        "Validation macro F1": f1_score(
            validation_classes,
            comparison_predictions,
            labels=[0, 1, 2],
            average="macro",
            zero_division=0
        )
    })

    scaler_loss_curves[scaler_name] = comparison_losses

scaler_results = pd.DataFrame(scaler_results).sort_values(
    "Validation macro F1", ascending=False
)

display(scaler_results)

for scaler_name, comparison_losses in scaler_loss_curves.items():
    plot.plot(
        range(1, 31),
        comparison_losses,
        label=scaler_name
    )

plot.xlabel("Training epoch")
plot.ylabel("Validation log loss")
plot.title("SGD convergence: StandardScaler versus MinMaxScaler")
plot.legend()
plot.show()

selected_scaler_name = scaler_results.iloc[0]["Scaler"]

selected_preprocessor = make_preprocessor(
    scaler_options[selected_scaler_name]
)

comparison_train = selected_preprocessor.fit_transform(
    train_features
)
comparison_test = selected_preprocessor.transform(test_features)

selected_classifier = SGDClassifier(
    loss="log_loss",
    penalty="l2",
    alpha=0.0001,
    learning_rate="constant",
    eta0=0.01,
    random_state=42
)

for epoch in range(30):
    selected_classifier.partial_fit(
        comparison_train,
        train_classes,
        classes=np.array([0, 1, 2])
    )

selected_predictions = selected_classifier.predict(comparison_test)
selected_probabilities = selected_classifier.predict_proba(
    comparison_test
)

print("Selected scaler:", selected_scaler_name)
print(
    "Selected scaler test accuracy:",
    accuracy_score(test_classes, selected_predictions)
)
print(
    "Selected scaler test macro F1:",
    f1_score(
        test_classes,
        selected_predictions,
        labels=[0, 1, 2],
        average="macro",
        zero_division=0
    )
)
print(
    "Selected scaler test log loss:",
    log_loss(
        test_classes,
        selected_probabilities,
        labels=[0, 1, 2]
    )
)
```

***Task 7***

Transition to Regression Tasks

```python
import matplotlib.pyplot as plot

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

regression_models = {"Linear regression": LinearRegression()}
for alpha_value in [0.01, 0.1, 1.0, 10.0, 100.0]:
    regression_models[f"Ridge alpha={alpha_value}"] = Ridge(alpha=alpha_value)
for alpha_value in [0.001, 0.01, 0.1, 1.0, 10.0]:
    regression_models[f"Lasso alpha={alpha_value}"] = Lasso(alpha=alpha_value, max_iter=20000)

regression_rows = []
coefficient_rows = []
for model_name, estimator in regression_models.items():
    model_pipeline = Pipeline([("prepare", make_preprocessor()), ("model", clone(estimator))])
    with warnings.catch_warnings(record=True) as warning_list:
        warnings.simplefilter("always", ConvergenceWarning)
        model_pipeline.fit(fit_features, fit_target)
    predictions = model_pipeline.predict(validation_features)
    coefficients = model_pipeline.named_steps["model"].coef_
    feature_names = model_pipeline.named_steps["prepare"].get_feature_names_out()
    model_family = "Linear" if model_name == "Linear regression" else model_name.split()[0]
    regression_rows.append({
        "Model": model_name, "Family": model_family,
        "Alpha": getattr(estimator, "alpha", 0.0),
        "Validation RMSE": np.sqrt(mean_squared_error(validation_target, predictions)),
        "Validation MAE": mean_absolute_error(validation_target, predictions),
        "Validation R2": r2_score(validation_target, predictions),
        "Coefficient L2 norm": np.linalg.norm(coefficients),
        "Zero coefficients": int(np.sum(np.abs(coefficients) < 1e-8)),
        "Convergence warning": any(issubclass(item.category, ConvergenceWarning) for item in warning_list)
    })
    for feature_name, coefficient in zip(feature_names, coefficients):
        coefficient_rows.append({"Model": model_name, "Feature": feature_name, "Coefficient": coefficient})

regression_results = pd.DataFrame(regression_rows).sort_values("Validation RMSE")
display(regression_results)
selected_rows = regression_results.groupby("Family", sort=False).head(1)
test_regression_rows = []
for model_name in selected_rows["Model"]:
    model_pipeline = Pipeline([
        ("prepare", make_preprocessor()), ("model", clone(regression_models[model_name]))
    ])
    model_pipeline.fit(train_features, train_target)
    predictions = model_pipeline.predict(test_features)
    test_regression_rows.append({
        "Model": model_name,
        "Test RMSE": np.sqrt(mean_squared_error(test_target, predictions)),
        "Test MAE": mean_absolute_error(test_target, predictions),
        "Test R2": r2_score(test_target, predictions)
    })
display(pd.DataFrame(test_regression_rows))
best_regression_name = regression_results.iloc[0]["Model"]

coefficient_table = pd.DataFrame(coefficient_rows).pivot(index="Feature", columns="Model", values="Coefficient")
display(coefficient_table)
for model_family in ["Ridge", "Lasso"]:
    family_results = regression_results[regression_results["Family"] == model_family].sort_values("Alpha")
    display(family_results[["Alpha", "Coefficient L2 norm", "Zero coefficients", "Validation RMSE"]])
    plot.plot(family_results["Alpha"], family_results["Coefficient L2 norm"], marker="o", label=model_family)
plot.xscale("log")
plot.xlabel("Regularization alpha")
plot.ylabel("Coefficient L2 norm")
plot.title("Effect of regularization strength")
plot.legend()
plot.show()

lasso_name = selected_rows.loc[selected_rows["Family"] == "Lasso", "Model"].iloc[0]
lasso_coefficients = coefficient_table[lasso_name]
```
